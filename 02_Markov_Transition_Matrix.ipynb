From a given input sequence, we can easily calculate a frequency table as well as a Markov transition matrix. This is done by counting the number of occurences of each nucleotide and also the occurence of each consecutive pair of nucleotides in the input sequence.

In a Markov transition matrix $P$, the matrix entry in the $i^{\text{th}}$ row and $j^{\text{th}}$ column is the probability of transition from state $i$ to state $j$, i.e. $P_{i, j} = \mathbb{P}(j\ |\ i)$. We can find this conditional probability experimentally from the given sequence.

$$P_{i, j} = \frac{\text{Number of transitions from state } i \text{ to state } j}{\text{Total number of transitions from state } i}$$

```python
# We will use the defaultdict library to create an empty dictionary
from collections import defaultdict

def build_frequency_table(sequence):
    # An initially empty dictionary to hold nucleotide counts
    frequency_table = defaultdict(int)

    # Count occurrences of each nucleotide
    for nucleotide in sequence:
        frequency_table[nucleotide] += 1

    return frequency_table

def build_markov_model(sequence):
    # As evident from the above formula, we need to count the number of each pairwise transition
    # and the total number of transitions from each state
    transition_counts = defaultdict(lambda: defaultdict(int))
    total_counts = defaultdict(int)

    # Count transitions by iterating through the sequence
    for i in range(len(sequence) - 1):
        current_state = sequence[i]
        next_state = sequence[i + 1]
        transition_counts[current_state][next_state] += 1
        total_counts[current_state] += 1

    # Build the transition matrix
    transition_matrix = {}
    for state, transitions in transition_counts.items():
        transition_matrix[state] = {next_state: count / total_counts[state] for next_state, count in transitions.items()}

    return transition_matrix
```

```python
# Driver code to test the functions

print("Enter a DNA sequence: ")
sequence = input().strip().upper()

# Currently the sequence entered is:
# TGGTCCACAGGACATTCGTCGCTTCCGGGTATGCGCTCTATGTGACGGTC

# Build the frequency table
frequency_table = build_frequency_table(sequence)
print("\nFrequency Table:")
for nucleotide, count in frequency_table.items():
    print(f"{nucleotide}: {count}")

# Build the Markov model
markov_model = build_markov_model(sequence)
print("\nMarkov Model:")
states = sorted(markov_model.keys())
print("     " + "      ".join(states))
for state in states:
    row = [f"{markov_model[state].get(next_state, 0):.3f}" for next_state in states]
    print(f"{state}: " + "  ".join(row))
```
