In this assignment, we will need to use basic Python file reading and string manipulation output. Note that we are reading from ```multiline_input.fasta``` and writing to a new file named as ```singleline_output.fasta```. The algorithm is simple and has been described in the comments.

````python
with open("data/multiline_input.fasta", "r") as infile:
    lines = infile.readlines() # Reads and stores the input file line by line

with open("data/singleline_output.fasta", "w") as outfile:
    seq = "" # Initializes an empty string that will always store the current sequence
    for line in lines:
        line = line.strip()
        if line.startswith(">"):
            if seq:
                outfile.write(seq + "\n") # Writes the current sequence to the output file
                seq = "" # Resets the sequence string for the next sequence
            outfile.write(line + "\n") # Writes the header line to the output file
        else:
            seq += line
    if seq: # If there is a sequence left at the end, write it to the output file
        outfile.write(seq + "\n")
````
